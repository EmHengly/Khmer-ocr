```python
import os
from transformers import PreTrainedTokenizerFast, PreTrainedTokenizer

# Define the save directory for tokenizer files
save_directory = "khmer_tokenizer"
os.makedirs(save_directory, exist_ok=True)

# Convert the JSON data to a list of (token, id) tuples
tokens = list(chartonum.keys())
ids = list(chartonum.values())

# Create a vocab.txt file from the JSON data
with open(os.path.join(save_directory, "vocab.txt"), 'w', encoding='utf-8') as f:
    for token, token_id in zip(tokens, ids):
        f.write(f"{token}\t{token_id}\n")

print("vocab.txt created successfully.")

# Create a fake SentencePiece model to satisfy PreTrainedTokenizerFast requirements
# The model file is required but can be a dummy one
dummy_spm_model_file = os.path.join(save_directory, "spiece.model")
with open(dummy_spm_model_file, 'w', encoding='utf-8') as f:
    f.write("")

# Define a dummy tokenizer class
class DummyTokenizer(PreTrainedTokenizer):
    def __init__(self, vocab_file, model_file):
        super().__init__()
        self.vocab_file = vocab_file
        self.model_file = model_file

    def get_vocab(self):
        return chartonum

# Create a tokenizer object
tokenizer = PreTrainedTokenizerFast(
    tokenizer_object=DummyTokenizer(vocab_file=os.path.join(save_directory, "vocab.txt"), model_file=dummy_spm_model_file),
    vocab_file=os.path.join(save_directory, "vocab.txt"),
    model_file=dummy_spm_model_file,
    unk_token='<unk>',
    pad_token='<pad>',
    eos_token='<eos>',
    bos_token='<bos>',
    add_special_tokens=True
)

# Save the tokenizer
tokenizer.save_pretrained(save_directory)

print("Tokenizer saved successfully.")
```
